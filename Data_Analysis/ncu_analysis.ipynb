```python
import pandas as pd
import matplotlib.pyplot as plt
import json
import os

# locate CSV files produced by ncu
report_dir = os.path.join(r"c:\UT\382V")
csv_files = [os.path.join(report_dir, f) for f in os.listdir(report_dir) if f.endswith('.csv')]
csv_files
```

```python
kernel_df = pd.read_csv(os.path.join(report_dir, 'Vision_Encoder_Data_transposed.csv'))
kernel_df['cumulative_time'] = kernel_df['gpu__time_duration.sum [us]'].cumsum()
plt.figure(figsize=(12,5))
plt.plot(kernel_df['cumulative_time'], kernel_df['dram__throughput.avg.pct_of_peak_sustained_elapsed [%]'], label='DRAM')
plt.plot(kernel_df['cumulative_time'], kernel_df['sm__throughput.avg.pct_of_peak_sustained_active [%]'], label='SM')
plt.xlabel('Time(us)')
plt.ylabel('Utlilization (% of peak)')
plt.title('Utilization During Vision Encoding')
plt.legend(loc='upper right', bbox_to_anchor=(1.15, 1))
plt.tight_layout()
plt.show()
```
